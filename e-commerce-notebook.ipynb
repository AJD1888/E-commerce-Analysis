```python
# function for creating dataframe only containing repeat customers
def repeat_customers_df(sales):
    return sales[sales["is_repeat_customer"] == True]

repeat_customers = repeat_customers_df(sales)

# function for creating dataframe only containing non repeat customers
def non_repeat_customers_df(sales):
    return sales[sales["is_repeat_customer"] == False]

non_repeat_customers = non_repeat_customers_df(sales)
```

```python
# function for total profits of returning and first time customers 
def total_profits(repeat_customers, non_repeat_customers):
     repeat_profits = np.round(repeat_customers["profit"].sum(),2)
     nonrepeat_profits = np.round(non_repeat_customers["profit"].sum(),2)

     return repeat_profits, nonrepeat_profits


repeat_profits, nonrepeat_profits = total_profits(repeat_customers, non_repeat_customers)

# display the two total profits
print("The total profit of returning customers is $" + str(repeat_profits))

print("The total profit of first time customers is $" + str(nonrepeat_profits))
```

```python
# we now want to compare the number of customers returning and the number of first time customers 

print("The number of returning customers in the dataset is " + str(repeat_customers.customer_id.nunique()))
print("The number of first time customers in the dataset is " + str(non_repeat_customers.customer_id.nunique()))
```

```python
# function to calculate the average profit per sale for each customer group

def average_profits(repeat_customers, non_repeat_customers):

    repeat_average = np.round(repeat_customers.profit.mean(),2)
    non_repeat_average = np.round(non_repeat_customers.profit.mean(),2)
    return repeat_average, non_repeat_average

repeat_avg, non_repeat_avg = average_profits(repeat_customers, non_repeat_customers)

# print the two average profits
print("The average profit per sale for a returning customer was $" + str(repeat_avg))
print("The average profit per sale for a first time cusotmer was $" + str(non_repeat_avg))
```

```python
# function for caculatung total profit maegin for returning and first time cusotmers 

def total_profit_margin(repeat_customers, non_repeat_customers):

    repeat_margin = np.round((repeat_customers.profit.sum() / repeat_customers.net_sales.sum()) * 100,1)

    non_repeat_margin = np.round((non_repeat_customers.profit.sum() / non_repeat_customers.net_sales.sum()) * 100,1)

    return repeat_margin, non_repeat_margin

repeat_margin, non_repeat_margin = total_profit_margin(repeat_customers, non_repeat_customers)

# display both margins on screen as a percentage 
print("The total profit margin for returning customers was " + str(repeat_margin) + "%")
print("The total profit margin for non returning customers was " + str(non_repeat_margin) + "%")
```

```python
# function to calculate the average profit per customer of returning and non returning customers 
def avg_prof_cust(repeat_customers, non_repeat_customers):

    avg1 = np.round(repeat_customers.profit.sum() / repeat_customers.customer_id.nunique(),2)
    avg2 = np.round(non_repeat_customers.profit.sum() / non_repeat_customers.customer_id.nunique(), 2)

    return avg1, avg2

profit_per_repeat, profit_per_non_repeat = avg_prof_cust(repeat_customers, non_repeat_customers)

# display the average profit per customer for both customer groups
print("The average profit per customer of returning customers is $" + str(profit_per_repeat))
print("The average profit per first time customer is $" + str(profit_per_non_repeat))
```

```python
# create a dataframe showing the number of orders per customer and the group they are in
customer_summary = sales.groupby("customer_id").agg(
    orders=("order_id", "nunique"),
    returning_customer=("is_repeat_customer", "first")
)

customer_summary.head()

```

```python
# we now want to get the average order frequency for each group of customer 
customer_summary.groupby("returning_customer")["orders"].mean().round(2)
```

```python
# we now want to calculate the average order per customer group

# create a dataframe storing all the import information about each order group
def create_orders_df(sales):

 orders_value = sales[["order_id", "customer_id", "is_repeat_customer", "net_sales"]]
 return orders_value

orders_value = create_orders_df(sales)
orders_value.head()
```

```python
# we now want to group our new dataframe by customer group and caculate the mean net sales value 

def average_order_value(orders_value):
 avg = orders_value.groupby("is_repeat_customer")["net_sales"].mean().round(2)
 return avg

avg_order_value = average_order_value(orders_value)
avg_order_value
```

```python
# we want to calculate pearson correlation for discount amount and net sales

discount_sales_corr, sales_p_value = pearsonr(sales["discount_amount"], sales["net_sales"])

print("Pearsons correlation is " + str(np.round(discount_sales_corr,2)) + " for discount amount and net sales")
```

```python
# calculate pearons correlation for discount amount and profit margin percenteage 

discount_profit_corr, profit_p_value = pearsonr(sales["discount_amount"], sales["profit_margin_percentage"])

print("Pearsons correlatuon is " + str(np.round(discount_profit_corr,2))+ " for discount amount and profit margin percentage")
```
