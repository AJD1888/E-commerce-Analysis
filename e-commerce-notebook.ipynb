```python
# function for creating dataframe only containing repeat customers
def repeat_customers_df(sales):
    return sales[sales["is_repeat_customer"] == True]

repeat_customers = repeat_customers_df(sales)

# function for creating dataframe only containing non repeat customers
def non_repeat_customers_df(sales):
    return sales[sales["is_repeat_customer"] == False]

non_repeat_customers = non_repeat_customers_df(sales)
```

```python
# function for total profits of returning and first time customers 
def total_profits(repeat_customers, non_repeat_customers):
     repeat_profits = np.round(repeat_customers["profit"].sum(),2)
     nonrepeat_profits = np.round(non_repeat_customers["profit"].sum(),2)

     return repeat_profits, nonrepeat_profits


repeat_profits, nonrepeat_profits = total_profits(repeat_customers, non_repeat_customers)

# display the two total profits
print("The total profit of returning customers is $" + str(repeat_profits))

print("The total profit of first time customers is $" + str(nonrepeat_profits))
```

```python
# we now want to compare the number of customers returning and the number of first time customers 

print("The number of returning customers in the dataset is " + str(repeat_customers.customer_id.nunique()))
print("The number of first time customers in the dataset is " + str(non_repeat_customers.customer_id.nunique()))
```

```python
# function to calculate the average profit per customer for each group

def average_profits(repeat_customers, non_repeat_customers):

    repeat_average = np.round(repeat_customers.profit.mean(),2)
    non_repeat_average = np.round(non_repeat_customers.profit.mean(),2)
    return repeat_average, non_repeat_average

repeat_avg, non_repeat_avg = average_profits(repeat_customers, non_repeat_customers)

# print the two average profits
print("The average profit per sale for a returning customer was $" + str(repeat_avg))
print("The average profit for a first time cusotmer was $" + str(non_repeat_avg))
```

```python
# function for caculatung total profit maegin for returning and first time cusotmers 

def total_profit_margin(repeat_customers, non_repeat_customers):

    repeat_margin = np.round((repeat_customers.profit.sum() / repeat_customers.net_sales.sum()) * 100,1)

    non_repeat_margin = np.round((non_repeat_customers.profit.sum() / non_repeat_customers.net_sales.sum()) * 100,1)

    return repeat_margin, non_repeat_margin

repeat_margin, non_repeat_margin = total_profit_margin(repeat_customers, non_repeat_customers)

# display both margins on screen as a percentage 
print("The total profit margin for returning customers was " + str(repeat_margin) + "%")
print("The total profit margin for non returning customers was " + str(non_repeat_margin) + "%")
```
